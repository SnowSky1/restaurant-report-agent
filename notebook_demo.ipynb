# 🏪 餐饮店经营分析智能体 - Notebook Demo

本 Notebook 演示如何在 Jupyter 环境中使用餐饮店经营分析智能体。

## 功能特点
- 🗺️ 位置分析 - 自动解析店铺地址
- 🏪 竞争对手分析 - 周边同类店铺分析
- 🚌 交通便利性 - 公交/地铁/停车场分布
- ⛅ 天气影响 - 天气对经营的影响
- 📊 交互式可视化 - Plotly 图表展示


## 1. 环境准备


```python
import sys
sys.path.insert(0, '.')

from dotenv import load_dotenv
load_dotenv()

from config.settings import settings
from tools.amap_tools import AmapTools
from tools.chart_tools import ChartTools
from graph.agent import create_agent_graph
from nodes.state import create_initial_state
from visualization.notebook import NotebookVisualizer

print("✓ 模块加载完成")
```

## 2. 配置店铺信息


```python
# 配置店铺信息 - 请修改为您要分析的店铺
STORE_NAME = "星巴克咖啡"
STORE_ADDRESS = "北京市朝阳区建国路88号SOHO现代城"
STORE_TYPE = "咖啡店"  # 可选: 餐厅、咖啡店、奶茶店、火锅店等
ANALYSIS_RADIUS = 1000  # 分析半径（米）

print(f"📍 店铺: {STORE_NAME}")
print(f"📍 地址: {STORE_ADDRESS}")
print(f"📍 类型: {STORE_TYPE}")
print(f"📍 分析半径: {ANALYSIS_RADIUS}米")
```

## 3. 初始化并运行分析


```python
# 初始化工具和工作流
amap_tools = AmapTools(mcp_url=settings.amap_mcp_url, api_key=settings.amap_maps_api_key)
chart_tools = ChartTools(mcp_url=settings.chart_mcp_url)
graph = create_agent_graph(amap_tools=amap_tools, chart_tools=chart_tools)

# 创建初始状态
initial_state = create_initial_state(
    store_name=STORE_NAME,
    store_address=STORE_ADDRESS,
    store_type=STORE_TYPE,
    analysis_radius=ANALYSIS_RADIUS
)

# 执行分析
print("🔄 正在分析，请稍候...")
final_state = await graph.ainvoke(initial_state)
print("✓ 分析完成!")
```

## 4. 可视化仪表板


```python
# 创建可视化器并显示完整仪表板
viz = NotebookVisualizer()
viz.display_analysis_dashboard(final_state)
```

## 5. 查看完整报告


```python
# 显示 Markdown 格式报告
viz.display_report(final_state.get('final_report', '报告生成失败'))
```

## 6. 清理资源


```python
await amap_tools.close()
await chart_tools.close()
print("✓ 资源已清理")
```
